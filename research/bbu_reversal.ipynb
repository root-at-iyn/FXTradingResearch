```python
pd.options.display.max_rows = 50
cols = ['Day_Idx', 'Open', 'High', 'Low', 'Close', 'Body', 'Range', 'ATR',
        'Close_%High', 'Open_%High', 'Iday_High', 'Iday_Low', 
        'Iday_Range', 'ADR', 'Close_%DHigh', 'Open_%DHigh', 
        'Yday_High', 'Yday_Low', 'SMA4', 'SMA16', 'SMA32',
       'BB_Upper_16_2', 'BB_Lower_16_2', 'RSI', 'Sig_High', 'Sig_Low',
       'SMA_Trend', 'SMA16_Slope', 'SMA32_Slope', 'Bear_BBR', 'Shooting_Star',
       'Bear_Engulf', 'Dark_Cloud', 'IHR', 'S/R', 'BBU_BO']
pd.options.display.max_columns = len(cols)
bbu_reversal = df[cols]#.query("Bear_BBR == True")
bbu_reversal
```

- Create new signal dataframe
- index is the signal id e.g. 0,1,2...
- columns = 
    - datetime of signal
    - max price up from start of signal until end of day
    - max price down from signal until start of day
    - lowest low from signal until EOD
    - lowest close from signal until EOD
    - highest high from signal until EOD
    - highest close from signal until EOD
    - Close at EOD
    

```python
signal = []
row = bbu_reversal.iloc[47]
s = {}
START = row.name
s["Date"] = START
s["Entry_Price"] = row["Close"]
day = row["Day_Idx"]
window = bbu_reversal.query(f"Day_Idx == {day}")
signal_window = window[START:]
s["Max_High"] = signal_window["High"].max() # Max high from signal until EOD
s["Min_Low"] = signal_window["Low"].min() # Min low from signal until EOD
s["Max_Up"] = s["Entry_Price"] - s["Max_High"] # Max price gain from signal until EOD
s["Max_Down"] = s["Entry_Price"] - s["Min_Low"] # Max price loss from signal until EOD
s["EOD_Close"] = signal_window.iloc[-1]["Close"] # Closing price at EOD
s["EOD_Gain"] = s["Entry_Price"] - s["EOD_Close"] # Gain at EOD from signal price


if signal_window.query(f"Close > {row["High"]}").empty is False:
    # If price closed above signal price within the day, 
    # get the dataframe between start and failure time
    EXIT = signal_window[START:].query(f"Close > {row["High"]}").iloc[0].name
    entry_window = signal_window[START:EXIT]
else:
    # If price didn't close above signal price within the day
    # get the dataframe from signal time until EOD
    EXIT = signal_window[START:].query(f"Close > {row["High"]}").iloc[0].name
    entry_window = signal_window

# If the price fell to the SMA16 after the signal
# calcuate the gain
below_sma16 = entry_window.query("Low < SMA16")
if below_sma16.empty is False:
    s["To_SMA16"] = s["Entry_Price"] - below_sma16.iloc[0]["SMA16"] 
else: 
    s["To_SMA16"] = 0

# If the price fell to the SMA32 after the signal
# calcuate the gain
below_sma32 = entry_window.query("Low < SMA32")
if below_sma32.empty is False:
    s["To_SMA32"] = s["Entry_Price"] - below_sma32.iloc[0]["SMA32"]
else:
    s["To_SMA32"] = 0

# If the price fell to the BB_Lower_16_2 after the signal
# calcuate the gain
below_bbl = entry_window.query("Low < BB_Lower_16_2")
if below_bbl.empty is False:
    s["To_BBL"] = s["Entry_Price"] - below_bbl.iloc[0]["BB_Lower_16_2"]
else:
    s["To_BBL"] = 0

s

```
